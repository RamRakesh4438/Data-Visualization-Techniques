```python
import pandas as pd

df = pd.read_csv("train.csv")

print("Dataset Shape:", df.shape)
df.head()
```

```python
print(df.info())
print("\nMissing values:")
print(df.isnull().sum())
print("\nDuplicates:", df.duplicated().sum())
```

```python
# Remove unnecessary index column
df = df.drop(columns=["Unnamed: 0"])

# Fill missing arrival delays with the median
df["Arrival Delay in Minutes"] = df["Arrival Delay in Minutes"].fillna(
    df["Arrival Delay in Minutes"].median()
)

print("Shape after cleaning:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Duplicates:", df.duplicated().sum())
```

```python
# Total delay
df["total_delay"] = (
    df["Departure Delay in Minutes"] +
    df["Arrival Delay in Minutes"]
)

# Age groups
df["age_group"] = pd.cut(
    df["Age"],
    bins=[0, 18, 30, 45, 60, 100],
    labels=["Under 18", "18-30", "31-45", "46-60", "60+"]
)

# Flight distance categories
df["distance_category"] = pd.cut(
    df["Flight Distance"],
    bins=[0, 1000, 2000, 4000, float("inf")],
    labels=["Short", "Medium", "Long", "Very Long"]
)

print(df[["Age", "age_group", "Flight Distance",
          "distance_category", "total_delay"]].head())
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("/", "_")
    .str.replace("-", "_")
)

print(df.columns.tolist())
```

```python
df.to_csv("airline_passenger_satisfaction_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
print("Final shape:", df.shape)
```
