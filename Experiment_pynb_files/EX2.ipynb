```python
# ============================================================
# EXPERIMENT NO: 2
# EDA - DATA IMPORT AND EXPORT
# ============================================================

# Import required libraries

import pandas as pd
import sqlite3

# Load the Diabetes Dataset

df = pd.read_csv(
    "diabetic_data.csv",
    na_values="?",
    low_memory=False
)

# Display complete dataset

print("===== DATASET =====")
print(df)

# Display first five records

print("\n===== FIRST FIVE ROWS =====")
print(df.head())

# Display last five records

print("\n===== LAST FIVE ROWS =====")
print(df.tail())

# Display dataset information

print("\n===== DATASET INFORMATION =====")
print(df.info())

# Display statistical summary

print("\n===== STATISTICAL SUMMARY =====")
print(df.describe(include="all"))

# Display dataset shape

print("\nDataset Shape:")
print(df.shape)

# Display column names

print("\n===== COLUMN NAMES =====")
print(df.columns)

# Check missing values

print("\n===== MISSING VALUES =====")
print(df.isnull().sum())

# Export dataset to CSV

df.to_csv(
    "diabetic_data_export.csv",
    index=False
)

print("\nCSV file exported successfully!")

# Export dataset to Excel

df.to_excel(
    "diabetic_data_export.xlsx",
    index=False
)

print("Excel file exported successfully!")

# Create SQLite Database

conn = sqlite3.connect(
    "diabetes_dataset.db"
)

# Export DataFrame to SQL Table

df.to_sql(
    "diabetic_data",
    conn,
    if_exists="replace",
    index=False
)

print("Dataset exported successfully to SQLite Database!")

conn.close()
```
