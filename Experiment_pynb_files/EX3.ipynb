```python
import pandas as pd

# Load Dataset
df = pd.read_csv("diabetic_data.csv", na_values="?", low_memory=False)

print("===== DATASET INFO =====")
df.info()

print("\n===== MISSING VALUES IN EACH COLUMN =====")
print(df.isnull().sum())

print("\n===== TOTAL MISSING VALUES =====")
print(df.isnull().sum().sum())

print("\n===== ANY MISSING VALUES? =====")
print(df.isnull().values.any())

print("\n===== NON-MISSING VALUES COUNT =====")
print(df.notnull().sum())
```

```python
import pandas as pd

df = pd.read_csv("diabetic_data.csv", na_values="?", low_memory=False)

# Fill all missing values with a constant value

df_constant = df.fillna("Unknown")
print("===== CONSTANT VALUE =====")
print(df_constant.head())

# Fill race column with mode

df_mode = df.copy()
df_mode["race"] = df_mode["race"].fillna(df_mode["race"].mode()[0])
print("\n===== MODE =====")
print(df_mode.head())

# Fill diagnosis columns with mode

df_diag = df.copy()

df_diag["diag_1"] = df_diag["diag_1"].fillna(df_diag["diag_1"].mode()[0])
df_diag["diag_2"] = df_diag["diag_2"].fillna(df_diag["diag_2"].mode()[0])
df_diag["diag_3"] = df_diag["diag_3"].fillna(df_diag["diag_3"].mode()[0])

print("\n===== DIAGNOSIS MODE =====")
print(df_diag.head())

# Forward Fill

df_ffill = df.ffill()
print("\n===== FORWARD FILL =====")
print(df_ffill.head())

# Backward Fill

df_bfill = df.bfill()
print("\n===== BACKWARD FILL =====")
print(df_bfill.head())
```

```python
import pandas as pd

df = pd.read_csv("diabetic_data.csv", na_values="?", low_memory=False)

# Drop rows with missing values

df_drop_rows = df.dropna()

print("===== DROP ROWS =====")
print(df_drop_rows.head())

# Drop columns with missing values

df_drop_cols = df.dropna(axis=1)

print("\n===== DROP COLUMNS =====")
print(df_drop_cols.head())

# Drop rows having any missing value

df_any = df.dropna(how="any")

print("\n===== DROP HOW='ANY' =====")
print(df_any.head())

# Drop rows where all values are missing

df_all = df.dropna(how="all")

print("\n===== DROP HOW='ALL' =====")
print(df_all.head())

# Keep rows with at least 40 non-null values

df_thresh = df.dropna(thresh=40)

print("\n===== THRESH=40 =====")
print(df_thresh.head())

# Drop rows where race or diag_1 is missing

df_subset = df.dropna(subset=["race", "diag_1"])

print("\n===== SUBSET (race, diag_1) =====")
print(df_subset.head())

# Drop missing values permanently

df.dropna(inplace=True)

print("\n===== AFTER inplace=True =====")
print(df.head())
```
