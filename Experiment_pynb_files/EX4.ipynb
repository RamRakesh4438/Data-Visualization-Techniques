```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Import Dataset
df = pd.read_csv("diabetic_data.csv", na_values="?", low_memory=False)

# ==========================
# Dataset Inspection
# ==========================

print("===== FIRST FIVE RECORDS =====")
display(df.head())

print("\n===== SHAPE OF DATASET =====")
print(df.shape)

print("\n===== DATA TYPES =====")
print(df.dtypes)

print("\n===== SUMMARY =====")
display(df.describe())

# ==========================
# Filtering Data
# ==========================

# Patients staying more than 7 days

print("\n===== PATIENTS STAYING MORE THAN 7 DAYS =====")

long_stay = df[df["time_in_hospital"] > 7]

display(long_stay.head())

# Female Patients

print("\n===== FEMALE PATIENTS =====")

female = df[df["gender"] == "Female"]

display(female.head())

# ==========================
# Sorting Data
# ==========================

print("\n===== SORT BY NUMBER OF MEDICATIONS (DESCENDING) =====")

sorted_df = df.sort_values(
    by="num_medications",
    ascending=False
)

display(sorted_df.head())

# ==========================
# Statistical Analysis
# ==========================

print("\n===== NUMBER OF MEDICATIONS STATISTICS =====")

print("Mean :", df["num_medications"].mean())

print("Median :", df["num_medications"].median())

print("Mode :")

print(df["num_medications"].mode())

print("Standard Deviation :", df["num_medications"].std())

print("\n===== DESCRIPTIVE STATISTICS =====")

display(df.describe())

# ==========================
# Grouping and Aggregation
# ==========================

print("\n===== AVERAGE HOSPITAL STAY BY GENDER =====")

gender_avg = df.groupby("gender")["time_in_hospital"].mean()

display(gender_avg)

# ==========================
# Correlation Analysis
# ==========================

print("\n===== CORRELATION =====")

corr = df[[
    "time_in_hospital",
    "num_lab_procedures",
    "num_procedures",
    "num_medications",
    "number_outpatient",
    "number_emergency",
    "number_inpatient",
    "number_diagnoses"
]].corr()

display(corr)

# ==========================
# Visualization 1
# ==========================

gender_avg.plot(kind="bar")

plt.title("Average Hospital Stay by Gender")

plt.xlabel("Gender")

plt.ylabel("Average Hospital Stay")

plt.show()

# ==========================
# Visualization 2
# ==========================

plt.figure(figsize=(8,5))

plt.scatter(
    df["num_lab_procedures"],
    df["num_medications"]
)

plt.title("Lab Procedures vs Number of Medications")

plt.xlabel("Lab Procedures")

plt.ylabel("Number of Medications")

plt.show()
```
